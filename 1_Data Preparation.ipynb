### 1. Data Cleaning

- Various sensor data (temperature, pressure, humidity, etc.) were collected during the manufacturing process.
- The data were cleaned by handling outliers and missing values, and removing duplicate records to prepare for analysis.

```python
# Import necessary libraries
import pandas as pd
import numpy as np

# Generate dummy manufacturing big data
np.random.seed(42) # Set random seed for reproducibility
data = {
    'Sensor_ID': np.arange(1, 21), # Sensor ID
    'Temperature': np.random.normal(75, 10, 20), # Temperature data (Normal distribution)
    'Pressure': np.random.normal(1000, 50, 20), # Pressure data (Normal distribution)
    'Humidity': np.random.uniform(30, 70, 20) # Humidity data (Uniform distribution)
}

# Create DataFrame
df = pd.DataFrame(data)

# Add some missing values and outliers
df.loc[5, 'Temperature'] = np.nan # Add missing value
df.loc[7, 'Pressure'] = np.nan # Add missing value
df.loc[9, 'Humidity'] = 120 # Outlier (Humidity exceeds the 0-100 range)

# Display DataFrame (Before cleaning)
print("\n Data before cleaning:")
print(df)
```

```python
# -- Data Cleaning -- #

# 1. Handle Missing Values: Replace with the mean value
df['Temperature'] = df['Temperature'].fillna(df['Temperature'].mean())
df['Pressure'] = df['Pressure'].fillna(df['Pressure'].mean())

# 2. Handle Outliers: Limit values to 0-100 range or replace with median (Humidity)
# Here, if value is outside 0-100, it is replaced by the median
df['Humidity'] = np.where((df['Humidity'] > 100) | (df['Humidity'] < 0), 
                          df['Humidity'].median(), 
                          df['Humidity'])

# 3. Remove Duplicate Data: Drop duplicate rows (Adding a dummy duplicate for demonstration)
df = pd.concat([df, df.iloc[[0]]], ignore_index=True) # Manually add the first row as a duplicate
df.drop_duplicates(inplace=True)

# Display DataFrame (After cleaning)
print("\n Data after cleaning:")
print(df)

print("\n Data Summary Statistics:")
print(df.describe())
```

### 2. Data Preprocessing

- In the manufacturing process, data is collected from various sensors—such as temperature, pressure, humidity, and machine status. 
- Before this data can be analyzed or used for machine learning, it must undergo a preprocessing stage to ensure accuracy and consistency. 
- The following example demonstrates the data preprocessing workflow using synthetic manufacturing data.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder

# Generate synthetic manufacturing big data
np.random.seed(42) # Set random seed for reproducibility
data = {
    'Sensor_ID': np.arange(1, 21), # Sensor ID
    'Temperature': np.random.normal(75, 10, 20), # Temperature data (Normal distribution)
    'Pressure': np.random.normal(1000, 50, 20), # Pressure data (Normal distribution)
    'Humidity': np.random.uniform(30, 70, 20), # Humidity data (Uniform distribution)
    'Status': np.random.choice(['Normal', 'Faulty'], size=20) # Sensor status (Categorical)
}

# Create DataFrame
df = pd.DataFrame(data)

# Add missing values and outliers
df.loc[5, 'Temperature'] = np.nan # Add missing value
df.loc[7, 'Pressure'] = np.nan # Add missing value
df.loc[9, 'Humidity'] = 120 # Add outlier (Humidity exceeds 0-100% range)

# Display DataFrame (Before preprocessing)
print("Data before preprocessing:")
print(df)
```

```python
# ----------- Data Preprocessing ----------- #

# 1. Handle Missing Values: Replace with the mean value
df['Temperature'] = df['Temperature'].fillna(df['Temperature'].mean())
df['Pressure'] = df['Pressure'].fillna(df['Pressure'].mean())

# 2. Handle Outliers: Restrict Humidity to 0-100 range
# If a value is outside the range, replace it with the median
df['Humidity'] = np.where((df['Humidity'] > 100) | (df['Humidity'] < 0), 
                          df['Humidity'].median(), 
                          df['Humidity'])

# 3. Categorical Data Encoding: Convert 'Status' column to numerical values
label_encoder = LabelEncoder()
df['Status_encoded'] = label_encoder.fit_transform(df['Status'])

# 4. Data Scaling: Standardize Temperature, Pressure, and Humidity
# This transforms data to have a mean of 0 and standard deviation of 1
scaler = StandardScaler()
df[['Temperature_scaled', 'Pressure_scaled', 'Humidity_scaled']] = scaler.fit_transform(
    df[['Temperature', 'Pressure', 'Humidity']]
)

# 5. Remove Duplicate Data: Add a dummy duplicate row and then remove it
df = pd.concat([df, df.iloc[[0]]], ignore_index=True) # Add first row as duplicate
df.drop_duplicates(inplace=True)

# Display DataFrame (After Preprocessing)
print("\n Data after preprocessing:")
print(df)

# Display Data Summary Statistics
print("\n Data Summary Statistics:")
print(df.describe())
```

### 3. Feature Engineering

- In the manufacturing process, Feature Engineering is the practice of creating new variables (features) or transforming existing ones based on collected sensor data. 
- This process is essential for improving the predictive performance of machine learning models and extracting deeper insights from the manufacturing environment.

```python
# Import necessary libraries
import pandas as pd
import numpy as np

# Generate synthetic manufacturing big data
np.random.seed(42)
data = {
    'Sensor_ID': np.arange(1, 21),
    'Temperature': np.random.normal(75, 10, 20), # Temperature (Normal Distribution)
    'Pressure': np.random.normal(1000, 50, 20),    # Pressure (Normal Distribution)
    'Humidity': np.random.uniform(30, 70, 20),      # Humidity (Uniform Distribution)
    'Production_Time': np.random.normal(30, 5, 20), # Production Time (Normal Distribution)
    'Status': np.random.choice(['Normal', 'Faulty'], size=20) # Machine Status
}

# Create DataFrame
df = pd.DataFrame(data)

# Display DataFrame (Before feature engineering)
print("Data before feature engineering:")
print(df)
```

```python
# ---------- Feature Engineering ---------- #

# 1. Ratio-based Feature: Ratio of Temperature to Pressure
# Useful for detecting efficiency imbalances in production equipment.
df['Temp_to_Pressure_Ratio'] = df['Temperature'] / df['Pressure']

# 2. Log Transformation: Log transform of Production Time
# Normalizes skewed distributions to improve model stability.
df['Log_Production_Time'] = np.log1p(df['Production_Time']) # Applied as log(1+x)

# 3. Categorical Encoding: Convert 'Status' column to numerical values
# Faulty = 1, Normal = 0 (Target encoding)
df['Status_encoded'] = df['Status'].apply(lambda x: 1 if x == 'Faulty' else 0)

# 4. Interaction Feature: Product of Temperature and Humidity
# Captures combined effects that individual features might miss.
df['Temp_Humidity_Interaction'] = df['Temperature'] * df['Humidity']

# 5. Moving Average: Rolling average to detect recent sensor trends
# Uses a window of 3 to smooth out noise and identify shifts.
df['Rolling_Temperature'] = df['Temperature'].rolling(window=3, min_periods=1).mean()

# 6. Binary Flag: Mark values above a specific threshold
# Creates a flag for 'High Temperature' (above 80).
df['High_Temperature_Flag'] = np.where(df['Temperature'] > 80, 1, 0)

# Display DataFrame (After Feature Engineering)
print("\n Data after feature engineering:")
print(df)

# Display Data Summary Statistics
print(df.describe())

# ---------- End of Feature Engineering ---------- #
```

### 4. Result Analysis
After performing feature engineering, we analyze the newly created variables to understand the underlying patterns in the manufacturing data. 
This step is crucial for identifying potential equipment failures and optimizing production efficiency.

- Correlation via Interaction: By creating the Temp_Humidity_Interaction feature, we can now see if the combination of high heat and high moisture leads to a 'Faulty' status more often than temperature alone.
- Trend Detection: The Rolling_Temperature smoothed out random sensor noise, allowing us to see if a machine's temperature was steadily rising before a failure occurred.
- Distribution Normalization: The Log_Production_Time transformation corrected the skewness of the original data, making it much easier for statistical models (like Linear Regression) to process the time data accurately.
