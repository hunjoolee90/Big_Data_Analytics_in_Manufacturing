# Data Architecture Design for Manufacturing Analytics

Designing a data architecture for manufacturing analytics is the process of structuring how data is collected, stored, processed, and analyzed from various sensors and systems. In this workflow, a pipeline is designed consisting of Data Collection (ETL), Storage (Database), and Processing & Analysis stages to ensure efficient data management and insights.

The following is an example of a complete data pipeline implemented using Python. It demonstrates collecting manufacturing data, storing it in a database, and then retrieving it for processing and analysis.

## Scenario

- The goal is to collect sensor data gathered from the manufacturing process from a CSV file, store it in an SQLite database, and then retrieve the data to perform data processing and analysis.
  - Data Collection (Collect from a CSV file)
  - Data Storage (Store in an SQLite database)
  - Data Processing & Analysis (Retrieve data from the database and perform analysis)

```python
import pandas as pd
import sqlite3
import numpy as np

#---- 1. Data Collection: Read manufacturing data from CSV file ----#
# Generating synthetic manufacturing sensor data (Simulating reading from an actual CSV file)

sensor_data = {
    'Timestamp': pd.date_range('2023-01-01', periods=100, freq='H'),
    'Temperature': np.random.normal(75, 5, 100),
    'Pressure': np.random.normal(1000, 50, 100),
    'Vibration': np.random.uniform(0.1, 1.0, 100),
    'Humidity': np.random.uniform(30, 70, 100),
    'Quality': np.random.choice(['Normal', 'Faulty'], 100, p=[0.9, 0.1])
}

# Convert to DataFrame
df = pd.DataFrame(sensor_data)

# Print manufacturing data (Collected data)
print("Collected Manufacturing Data:")
print(df.head())


#---- 2. Data Storage: Save to SQLite database ----#

# Connect to SQLite database (Create database in memory or as a file)
conn = sqlite3.connect('manufacturing_data.db')

# Store data by converting DataFrame to SQL table
df.to_sql('sensor_data', conn, if_exists='replace', index=False)

print("\nData has been saved to the SQLite database.")


#---- 3. Data Processing & Analysis: Retrieve from database and analyze ----#

# Retrieve data from SQLite database
query = "SELECT * FROM sensor_data WHERE Quality = 'Faulty'"
df_faulty = pd.read_sql(query, conn)

# Print only 'Faulty' data from the stored records
print("\nFaulty Data Analysis:")
print(df_faulty.head())

# Data Analysis: Calculate mean sensor values for faulty products
mean_values = df_faulty[['Temperature', 'Pressure', 'Vibration', 'Humidity']].mean()

print("\nMean sensor values for faulty products:")
print(mean_values)

# Close database connection
conn.close()
```

## Result Analysis

### 1. Data Ingestion & Storage
* **Collection**: Successfully gathered 100 hourly logs from manufacturing sensors.
* **Storage**: Data was migrated to the `manufacturing_data.db` SQLite database for structured management.
